```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)

population = np.random.randint(18, 61, 200)

population[:10]
```

```python
sample_size = 20

sample = np.random.choice(population, sample_size, replace=False)

sample
```

```python
plt.figure()

plt.hist(population, alpha=0.6, label="Full Population")
plt.hist(sample, alpha=0.8, label="Random Sample")

plt.legend()
plt.title("Simple Random Sampling")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
class1 = np.random.normal(65,5,30)
class2 = np.random.normal(75,5,30)
class3 = np.random.normal(85,5,30)

data = np.concatenate([class1, class2, class3])

labels = np.array(
    ["Class A"]*30 + ["Class B"]*30 + ["Class C"]*30
)

df = pd.DataFrame({"Score": data, "Class": labels})
df.head()
```

```python
stratified_sample = df.groupby("Class").sample(5)

stratified_sample
```

```python
plt.figure(figsize=(6,8))

plt.subplot(2,1,1)
df.boxplot(column="Score", by="Class")
plt.title("Original Data Distribution")

plt.subplot(2,1,2)
stratified_sample.boxplot(column="Score", by="Class")
plt.title("Stratified Sample Distribution")

plt.tight_layout()
plt.show()
```
