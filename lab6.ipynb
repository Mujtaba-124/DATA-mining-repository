```python
# Transactions Dataset
transactions = [
    (['Bread', 'Milk'], 'Buy'),
    (['Bread', 'Diaper', 'Beer', 'Eggs'], 'Buy'),
    (['Milk', 'Diaper', 'Beer', 'Cola'], 'NoBuy'),
    (['Bread', 'Milk', 'Diaper', 'Beer'], 'Buy'),
    (['Bread', 'Milk', 'Diaper', 'Cola'], 'Buy')
]

print("Dataset:")
for t in transactions:
    print(t)
```

```python
# Extract all items
all_items = []

for items, label in transactions:
    all_items.extend(items)

unique_items = list(set(all_items))

print("\nUnique Items:")
print(unique_items)
```

```python
total_transactions = len(transactions)

item_support = {}

for item in unique_items:
    count = 0
    for items, label in transactions:
        if item in items:
            count += 1

    item_support[item] = count / total_transactions

print("\nSupport for each item:")
for item, support in item_support.items():
    print(f"{item}: {support:.2f}")
```

```python
antecedent = "Bread"
consequent = "Buy"

freq_antecedent = 0
freq_both = 0

for items, label in transactions:
    if antecedent in items:
        freq_antecedent += 1
        if label == consequent:
            freq_both += 1

# Metrics
support = freq_both / total_transactions
confidence = freq_both / freq_antecedent

support_consequent = sum(
    1 for _, label in transactions if label == consequent
) / total_transactions

lift = confidence / support_consequent

print("\nAssociation Rule: IF Bread THEN Buy")
print(f"Support: {support:.2f}")
print(f"Confidence: {confidence:.2f}")
print(f"Lift: {lift:.2f}")
```

```python
antecedent = "Laptop"
consequent = "Headphones"

freq_antecedent = 0
freq_both = 0

for items, label in transactions:
    if antecedent in items:
        freq_antecedent += 1
        if consequent in items:
            freq_both += 1

# Support
support = freq_both / total_transactions

# Confidence
confidence = freq_both / freq_antecedent if freq_antecedent != 0 else 0

support_consequent = sum(
    1 for items, _ in transactions if consequent in items
) / total_transactions

# Lift
lift = confidence / support_consequent if support_consequent != 0 else 0

print("\nAssociation Rule: IF Laptop THEN Headphones")
print(f"Support: {support:.2f}")
print(f"Confidence: {confidence:.2f}")
print(f"Lift: {lift:.2f}")
```
