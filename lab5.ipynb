```python
import re

# Sample Text (you can replace with your own paragraph)
textData = """
Python is a powerful tool! It is widely used in data analysis,
machine learning, AI, and signal processing in 2025.
"""

print("Original Text:")
print(textData)

# Lowercase
cleanedText = textData.lower()

# Remove punctuation & special characters
cleanedText = re.sub(r'[^a-zA-Z\s]', '', cleanedText)

# Remove extra spaces
cleanedText = re.sub(r'\s+', ' ', cleanedText).strip()

print("\nCleaned Text:")
print(cleanedText)
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('stopwords')

# Tokenization
tokens = word_tokenize(cleanedText)

# Stopwords
stop_words = set(stopwords.words('english'))

filtered_tokens = [word for word in tokens if word not in stop_words]

print("\nTokens after Stopword Removal:")
print(filtered_tokens)
```

```python
import nltk

nltk.download('averaged_perceptron_tagger_eng')
```

```python
from nltk.stem import PorterStemmer
from nltk import pos_tag

nltk.download('averaged_perceptron_tagger')

# Stemming
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_tokens]

print("\nStemmed Words:")
print(stemmed_words)

# POS Tagging
pos_tags = pos_tag(filtered_tokens)

print("\nPOS Tags:")
print(pos_tags)
```

```python
from nltk.stem import PorterStemmer
from nltk import pos_tag

nltk.download('averaged_perceptron_tagger')

# Stemming
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_tokens]

print("\nStemmed Words:")
print(stemmed_words)

# POS Tagging
pos_tags = pos_tag(filtered_tokens)

print("\nPOS Tags:")
print(pos_tags)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

vectorizer = CountVectorizer()
X = vectorizer.fit_transform([' '.join(filtered_tokens)])

bow_df = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out()
)

print("\nBag of Words:")
print(bow_df)
```

```python
import matplotlib.pyplot as plt

word_counts = bow_df.iloc[0]

word_counts.sort_values(ascending=False).plot(kind='bar')

plt.title("Top Words Frequency")
plt.ylabel("Word Count")
plt.xticks(rotation=45)
plt.show()
```

```python
from wordcloud import WordCloud

wordcloud = WordCloud(width=800, height=400).generate(' '.join(filtered_tokens))

plt.figure(figsize=(10,5))
plt.imshow(wordcloud)
plt.axis('off')
plt.title("Word Cloud Visualization")
plt.show()
```
